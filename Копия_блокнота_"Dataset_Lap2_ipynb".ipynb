Написать скрипт, содержащий функцию, получающую на входе метку класса и возвращающую следующий экземпляр (путь к нему) этого класса

```python
class DataFetcher:
    def __init__(self, dataframe, job_title):
        self.data = dataframe[dataframe['job title'] == job_title].sample(frac = 1).reset_index(drop = True)
        self.index = 0

    def next_item(self):
        if self.index < len(self.data):
            row = self.data.iloc[self.index]
            self.index += 1
            return row
        else:
            return None

# # Пример
fetcher = DataFetcher(DataFrame,'Python Instructor for Kids 13-17 - Alexandria' )
print(fetcher.next_item())
print(fetcher.next_item())
```

Классы Итераторы

```python
# классы итераторы
class JobIterator:
    def __init__(self, dataframe, job_title):
        self.data = dataframe[dataframe['job title'] == job_title].sample(frac = 1).reset_index(drop = True)
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index < len(self.data):
            row = self.data.iloc[self.index]
            self.index += 1
            return row
        else:
            raise StopIteration

# Пример
iterator = JobIterator(DataFrame, 'Python Instructor for Kids 13-17 - Alexandria')
for item in iterator:
    print(item)

```
